Write the Python Program to solve 8-Puzzle
Problem.

```python
import heapq

goal_state = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def heuristic(state):
    distance = 0
    for i in range(3):
        for j in range(3):
            tile = state[i * 3 + j]
            if tile != 0:
                x, y = (tile - 1) // 3, (tile - 1) % 3
                distance += abs(x - i) + abs(y - j)
    return distance

def solve(initial_state):

    frontier = [(heuristic(initial_state), initial_state)]
    g_score_map = {initial_state: 0}

    f_score_map = {initial_state: heuristic(initial_state)}

    while frontier:
        current_f_score, current_state = heapq.heappop(frontier)
        if current_state in g_score_map and current_f_score > f_score_map[current_state]:
            continue

        if current_state == goal_state:
            return True

        for successor in successors(current_state):
            tentative_g_score = g_score_map.get(current_state, float('inf')) + 1
            if tentative_g_score < g_score_map.get(successor, float('inf')):
                g_score_map[successor] = tentative_g_score
                f_score_successor = tentative_g_score + heuristic(successor)
                f_score_map[successor] = f_score_successor
                heapq.heappush(frontier, (f_score_successor, successor))

    return False

def successors(state):
    successors = []
    i = state.index(0)
    if i % 3 != 0:
        new_state = list(state)
        new_state[i], new_state[i - 1] = new_state[i - 1], new_state[i]
        successors.append(tuple(new_state))

    if i % 3 != 2:
        new_state = list(state)
        new_state[i], new_state[i + 1] = new_state[i + 1], new_state[i]
        successors.append(tuple(new_state))

    if i // 3 != 0:
        new_state = list(state)
        new_state[i], new_state[i - 3] = new_state[i - 3], new_state[i]
        successors.append(tuple(new_state))

    if i // 3 != 2:
        new_state = list(state)
        new_state[i], new_state[i + 3] = new_state[i + 3], new_state[i]
        successors.append(tuple(new_state))
    return successors

initial_state = (2, 8, 3, 1, 6, 4, 7, 0, 5)
if solve(initial_state):
    print("The puzzle is solvable!")
else:
    print("The puzzle is unsolvable.")
```
